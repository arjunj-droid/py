```python
import pandas as pd
india_weather = pd.DataFrame({
    "city":['mumbai','delhi','bangalore'],
    "temperature":[32,45,30],
    "humidity":[80,60,78]
})
india_weather
    
```

```python
import pandas as pd
us_weather = pd.DataFrame({
    "city":['new york','chicago','orlando'],
    "temperature":[21,12,35],
    "humidity":[68,65,75]
})
us_weather
```

```python
#join this 2 
df = pd.concat([india_weather, us_weather])
df
#u can see the indexes are same 
```

```python
#to make index series
df = pd.concat([india_weather, us_weather],ignore_index=True)
df
```

```python
#you can also pass keys
#to make index series
df = pd.concat([india_weather, us_weather],keys=['india','us'])
df
```

```python
#if you want to retrieve specific country weather
df.loc['india']
```

```python
df.loc['us']
```

```python

temperature_df = pd.DataFrame({
    "city":['mumbai','delhi','banglore'],
    "temperature":[32,40,30],
})
temperature_df
```

```python
windspeed_df = pd.DataFrame({
    "city":['mumbai','delhi','banglore'],
    "windspeed":[7,12,9],
})
windspeed_df
```

```python
df = pd.concat([temperature_df, windspeed_df])
df
#it is creating additional rows windspeed should appear as column
```

```python
#axis=0 -->append second Dataframe as rows
#axis=1 -->append second dataframe as columns
df = pd.concat([temperature_df, windspeed_df],axis=1)
df
```

```python
temperature_df = pd.DataFrame({
    "city":['mumbai','delhi','banglore'],
    "temperature":[32,40,30],
},index=[0,1,2])
temperature_df
```

```python
#if you missing any value or inter changed values it will execute wrong data so to avoid that use index
windspeed_df = pd.DataFrame({
    "city":['delhi','mumbai'],
    "windspeed":[12,7],
},index=[1,0])
windspeed_df
```

```python
df = pd.concat([temperature_df, windspeed_df],axis=1)
df
```

```python
#join dataframe with series

s=pd.Series(['Humid','Dry','Rain'],name='event')
s
```

```python
#append this series to temperature_df dataframe
df=pd.concat([temperature_df,s],axis=1)
df
```
