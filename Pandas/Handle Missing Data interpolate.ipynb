```python
import pandas as pd
df = pd.read_csv("weather_data (2).csv")
df
```

```python
#interpolate() is used to fill missing values (NaN) by estimating values from existing data instead of using a fixed value like 0 or the column average.
new_df=df.fillna(0)
new_df
```

```python
new_df.dtypes
```

```python
#interpolate() is used to fill missing values (NaN) by estimating values from existing data instead of using a fixed value like 0 or the column average.
new_df = df.copy() #Create a copy df      -> Original and  new_df  -> Copy

new_df['day'] = pd.to_datetime(new_df['day']) #Convert string dates to datetime

new_df.set_index('day', inplace=True) #Make day the index, Because method='time' uses the dates in the index to calculate missing values. 

#Select the columns
new_df[['temperature', 'windspeed']] = (
    new_df[['temperature', 'windspeed']].interpolate(method='time') #Apply interpolation
)

new_df
```

```python
#it can be written also like this.
new_df = df.copy() 
new_df['day'] = pd.to_datetime(new_df['day'])
new_df.set_index('day', inplace=True)
new_df['temperature'] = new_df['temperature'].interpolate(method='time')
new_df['windspeed'] = new_df['windspeed'].interpolate(method='time')
new_df
```
