```python
import pandas as pd
df = pd.read_csv("weather_data (2).csv")
df
```

```python
#dropna() is used to remove rows or columns that contain missing values (NaN).
new_df = df.dropna()
new_df
```

```python
#Drop a row (or column) only if ALL values are NaN.
new_df = df.dropna(how='all')
new_df
```

```python
#thresh means minimum number of non-NaN values required to keep a row (or column).
#Drop a row (or column) only if ALL values are NaN.
new_df = df.dropna(thresh=1)#min 1 valid values
new_df
```

```python
new_df = df.dropna(thresh=3)#min 1 valid values
new_df
```

```python
#insert missing date
#reindex() → Insert missing dates and fill their values with NaN
#date_range() → creates all dates between start and end.
new_df = df.copy() #Create a copy

new_df['day'] = pd.to_datetime(new_df['day']) #Convert day to datetime

new_df = new_df.set_index('day') #Make day the index

new_df = new_df.reindex(
    pd.date_range(
        start='2017-01-01',
        end='2017-01-10'
    ) #Create all dates from Jan 1 to Jan 10
)

new_df
```
