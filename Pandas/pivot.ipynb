```python
#pivot table is used to transform and summarize data, making it easier to analyze relationships between columns.
df.pivot(index='date',columns='city')
```

```python
import pandas as pd
df=pd.read_csv('weather_data (5).csv')
df
#same day 2 diff temp and humid
```

```python
#create average temp and humid
#pivot() = No duplicates allowed.
#pivot_table() = Duplicates allowed, aggregates values.

df.pivot_table(index='city', columns='date')
```

```python
df.pivot_table(index='city', columns='date',aggfunc='sum')
```

```python
#margins=True adds totals (Grand Total) rows and/or columns to a pivot table.
#margins_name='Total' → changes the label from All to Total.
df.pivot_table(index='city', columns='date',margins=True)
```

```python
#avg tem month may and desc
#pd.Grouper() is used with groupby() or pivot_table() to group datetime data into time periods such as days, months, quarters, or years.
import pandas as pd

df = pd.read_csv("weather_data (5).csv")

df['date'] = pd.to_datetime(df['date'])

df.pivot_table(
    index=pd.Grouper(key='date', freq='ME'),
    columns='city'
)
```
