```python
import pandas as pd
df = pd.read_csv("weather_data (2).csv",parse_dates=['day'])
df.set_index('day', inplace=True)
df
```

```python
#Replace NaN with 0
new_df = df.fillna(0)
new_df
```

```python
#if you want to specift different values for diferent columns :(that time use dictionary)
new_df=df.fillna({
    'temperature':0,
    'windspeed':0,
    'event':'no event'})
new_df
```

```python
#if your calculating mean or something for temperature then ,mean is going to come horrible and someone looks data is everything ok they might notice sudden drop temp to 0 ,actually we dont know the temp on that day
#other way of doing estimate would be just to carry forward the previous day temperature and we can do similar way to other data type also
#ffill Fill values by propagating the last valid observation to next valid.
new_df = df.ffill()
new_df
```

```python
#we can also use backward fill meaning carry forward next day data
#pandas.DataFrame.fillna --goto this site and gett all fillna methods
#bfill Fill values by using the next valid observation to fill the gap.
new_df = df.bfill()
new_df
```

```python
#limit controls how many consecutive NaN values are filled.if we have 2 nan if limit set at 1 only 1 nan get filled
new_df = df.ffill(limit=1)
new_df
```
