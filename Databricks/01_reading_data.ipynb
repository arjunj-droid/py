```python
#reading the csv file, 1.In Spark, .option() expects two arguments: key and value.2.the option name should be lowercase inferSchema.
df = spark.read.format('csv').option('inferSchema', 'true').option('header', 'true').load('/Volumes/workspace/default/raw/sales_data.csv')
```
